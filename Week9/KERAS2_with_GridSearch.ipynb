```python
!pip install scikeras

import keras
from keras.models import Sequential
from keras.layers import Dense
import numpy as np

from scikeras.wrappers import KerasClassifier
from sklearn.model_selection import GridSearchCV
import pandas as pd
```

```python
def create_network(x=16,optimizer="rmsprop"):
    #Define the model achitecture
    model = Sequential()
    model.add(Dense(x, input_shape=(784,),activation='relu'))
    #model.add(Dropout(0.2))
    model.add(Dense(x,activation='relu'))
    #model.add(Dropout(0.2))
    #model.add(Dropout(0.2))
    model.add(Dense(10,activation='softmax'))
    model.compile(loss='categorical_crossentropy', optimizer=optimizer, metrics=["accuracy"])
    model.summary()
    return model
#Make the model learn
#MORE OPTIMIZER DETAILS here https://keras.io/optimizers/
```

```python
from google.colab import drive
drive.mount("/content/drive",force_remount=True)
path="/content/drive/My Drive/data/digit.csv"
df =pd.read_csv(path)
data=pd.read_csv(path,header=None)
```

```python

X=data.values

from sklearn.model_selection import train_test_split
X_train , X_test, Y_train, Y_test=train_test_split(X[:,:784]/256,X[:,784],test_size=0.2)
y_train = keras.utils.to_categorical(Y_train, num_classes=10)
y_test = keras.utils.to_categorical(Y_test, num_classes=10)
```

```python
network = KerasClassifier(build_fn=create_network, verbose=1)

param_grid = dict(
    epochs=[5, 10],
    batch_size=[10, 50],
    model__optimizer=["rmsprop", "adam"],
    model__x=[16, 32,64],
)

grid = GridSearchCV(estimator=network, param_grid=param_grid, n_jobs=-1,cv=3)
grid_result = grid.fit(X_train, y_train)
```

```python
print("Best: %f using %s" % (grid_result.best_score_, grid_result.best_params_))
means = grid_result.cv_results_['mean_test_score']
stds = grid_result.cv_results_['std_test_score']
params = grid_result.cv_results_['params']
for mean, stdev, param in zip(means, stds, params):
    print("%f (%f) with: %r" % (mean, stdev, param))
```
